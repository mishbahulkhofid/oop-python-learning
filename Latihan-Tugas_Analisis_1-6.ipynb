```python
class Hewan:
    def __init__(self, nama):
        self.nama = nama

    def makan(self):
        return f"{self.nama} sedang makan"

    def tidur(self):
        return f"{self.nama} sedang tidur"

    def bunyi(self):
        return f"{self.nama} membuat suara"


class Kucing(Hewan):  # <- Kucing mewarisi dari Hewan
    def bunyi(self):  # <- override method bunyi() milik Hewan
        return f"{self.nama}: Meong!"


class Anjing(Hewan):  # <- Anjing juga mewarisi dari Hewan
    def bunyi(self):
        return f"{self.nama}: Guk guk!"


kucing1 = Kucing("Milo")
anjing1 = Anjing("Rocky")

print(kucing1.makan())   # Milo sedang makan   <- method dari Hewan, otomatis kepakai
print(kucing1.bunyi())   # Milo: Meong!         <- method sendiri (override)

print(anjing1.tidur())   # Rocky sedang tidur   <- method dari Hewan
print(anjing1.bunyi())   # Rocky: Guk guk!      <- method sendiri
```

```python
class AkunBank:
    def __init__(self, saldo):
        self.__saldo = saldo # Private attribute

    def cek_saldo(self):
        return self.__saldo

    def setor(self, jumlah):
        if jumlah > 0:
            self.__saldo += jumlah

    def tarik(self, jumlah):
        if 0 < jumlah <= self.__saldo:
            self.__saldo -= jumlah
        else:
            print("Saldo tidak mencukupi")

rekening = AkunBank(1000)
rekening.tarik(1000)
rekening.setor(500)
print(rekening.cek_saldo()) # Output: 1500
# print(rekening.__saldo) # Ini akan Error! Tidak bisa akses  langsung.
```

*Parameter lawan adalah objek utuh agar program bisa memanggil fungsi lawan.diserang() dan mengurangi nilai hp lawan secara langsung. Jika hanya string nama, status HP lawan tidak akan bisa diperbarui*

```python
#Latihan 3

# Class Parent (Hero)
class Hero:

    def __init__(self, name, hp, attack_power):
        self.name = name
        self.hp = hp
        self.attack_power = attack_power

    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.nam}!")
        lawan.diserang(self.attack_power)

    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")


# Class Mage adalah anak dari class Hero
class Mage(Hero):

    def __init__(self, name, hp, attack_power, mana):
        # Memanggil constructor milik Parent (Hero)
        super().__init__(name, hp, attack_power)
        self.mana = mana

    def info(self):
        # Perbaikan: String dibuat dalam satu baris utuh
        print(f"{self.name} [Mage] | HP: {self.hp} | Mana: {self.mana}")

    # Mage punya skill khusus
    def skill_fireball(self, lawan):
        if self.mana >= 20:
            # : String dibuat dalam satu baris utuh
            print(f"{self.name} menggunakan Fireball ke {lawan.name}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2)  # Damage 2x lipat
        else:
            print(f"{self.name} gagal skill! Mana tidak cukup.")


# -- Main Program --
print("\n--- Update Class Hero ---")
eudora = Mage("Eudora", 80, 30, 100)
balmond = Hero("Balmond", 200, 10)

eudora.info()
eudora.serang(balmond)  # Serangan biasa (warisan dari Hero)
eudora.skill_fireball(balmond)  # Skill khusus Mage
```

```python
#Latihan 4

class Hero:
    def __init__(self, nama, hp_awal):
        self.nama = nama
        # Enkapsulasi: HP bersifat Private (hanya bisa diakses di
        # dalam class ini)
        self.__hp = hp_awal

    # GETTER: Cara resmi melihat HP
    def get_hp(self):
        return self.__hp

    # SETTER: Cara resmi mengubah HP (dengan validasi)
    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0  # HP tidak boleh negatif
        elif nilai_baru > 1000:
            print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

    def diserang(self, damage):
        # Kita pakai setter/getter bahkan di dalam class sendiri agar
        # aman
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)
        print(f"{self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}")

# -- Uji Coba --
hero1 = Hero("Layla", 100)
# hero1.__hp = 9999 # GAGAL (Tidak akan mengubah hp asli)
# print(hero1.__hp) # ERROR (Tidak bisa dibaca langsung)
hero1.set_hp(-50) # Coba set negatif
print(hero1.get_hp()) # Output: 0 (Karena dicegat oleh logika Setter)
```

1. Name Mangling

    Hasil: Nilai HP tetap muncul.

    Alasan: Python tidak benar-benar mengunci variabel private, melainkan hanya mengubah namanya menjadi _Hero__hp (Name Mangling).

    Bahaya: Mengaksesnya paksa merusak aturan enkapsulasi dan berisiko memicu bug karena menerobos sistem keamanan internal kelas.

2. Uji Validasi

    Hasil: HP Hero menjadi -100.

    Pentingnya Setter: Tanpa validasi Setter, aturan game rusak (HP bernilai negatif). Setter wajib ada sebagai filter otomatis agar nilai data selalu valid dan aman dari data sampah.

```python
# Parent Class
class Hero:
    def __init__(self, nama):
        self.nama = nama

    def serang(self):
        print("Hero menyerang dengan tangan kosong.")


# Child Class 1
class Mage(Hero):
    def serang(self):
        print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")


# Child Class 2
class Archer(Hero):
    def serang(self):
        print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")


# Child Class 3
class Fighter(Hero):
    def serang(self):
        print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")


# -- Penerapan Polymorphism --
# Kita punya daftar hero campuran
pasukan = [
    Mage("Eudora"),
    Archer("Miya"),
    Fighter("Zilong"),
    Mage("Gord")
]

print("--- PERANG DIMULAI ---")
# Satu perintah loop, tapi respon berbeda-beda (Polymorphism)
for pahlawan in pasukan:
    pahlawan.serang()
```

1. Error hapus serang() di Hero:
TypeError: Can't instantiate abstract class Hero without an implementation for abstract method 'serang' — Python menolak bikin objek Hero karena belum memenuhi janji dari GameUnit (wajib punya serang).

Konsekuensi lupa implementasi: program langsung error saat objek dibuat (bukan nanti pas dipanggil), jadi bug cepat ketahuan dan konsistensi desain tetap terjaga.

2. Kenapa GameUnit() dilarang:
Karena method-nya kosong (pass), tidak ada logika nyata untuk dijalankan.

Gunanya GameUnit: jadi cetakan/kontrak wajib buat semua class anak, supaya Hero, Monster, dll pasti punya serang() dan info() — dasar dari polymorphism.

```python
# Parent Class
class Hero:
    def __init__(self, nama):
        self.nama = nama

    def serang(self):
        print("Hero menyerang dengan tangan kosong.")


# Child Class 1
class Mage(Hero):
    def serang(self):
        print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")


# Child Class 2
class Archer(Hero):
    def serang(self):
        print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")


# Child Class 3
class Fighter(Hero):
    def serang(self):
        print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")

# Child Class 4 (Tugas Analisis)
class Healer  (Hero):
    def serang(self):
        print(f"{self.nama} (Healer) Tidak menyerang Tapi Menyembuhkan Teman")

# -- Penerapan Polymorphism --
pasukan = [
    Mage("Eudora"),
    Archer("Miya"),
    Fighter("Zilong"),
    Mage("Gord"),
    Healer("Rafaela")
]

print("--- PERANG DIMULAI ---")
for pahlawan in pasukan:
    pahlawan.serang()
```

1. Tambah class Healer:
Program tetap jalan lancar tanpa error, Healer otomatis ikut diproses loop.
Keuntungan Polymorphism: bisa nambah karakter baru tanpa ubah kode lama (loop) — scalable, minim bug.

2. Ganti nama method Archer jadi tembak_panah:
Tidak error, tapi salah logika — Archer malah pakai serang() default dari Hero ("menyerang dengan tangan kosong"), bukan versi memanah miliknya.
Kenapa nama harus sama: karena Polymorphism kerja lewat override — kalau nama beda, Python anggap itu method baru, bukan override, jadi mekanisme "satu perintah beda respons" rusak.
